```python
import sys, subprocess, pathlib, urllib.request, zipfile, hashlib, json, shutil, os
ROOT = pathlib.Path('/kaggle/working/Typed-decision-reliability-Research-Study')
def run(args, cwd=None):
    print('EXEC', ' '.join(map(str,args)), flush=True)
    subprocess.run(args, cwd=cwd, check=True)
if not ROOT.exists():
    run(['git','clone','https://github.com/XDR-SAM/Typed-decision-reliability-Research-Study.git',str(ROOT)])
run(['git','checkout','3d3b711c41259eea4433d6714fd2002bc68b99b0'],ROOT)
PROJECT = ROOT / 'research-project'
assets = json.loads((ROOT/'neural_artifact_manifest.json').read_text())['assets'][0]
archive = pathlib.Path('/kaggle/temp/neural-preparation-data.zip')
archive.parent.mkdir(parents=True,exist_ok=True)
if not archive.exists():
    urllib.request.urlretrieve('https://github.com/XDR-SAM/Typed-decision-reliability-Research-Study/releases/download/neural-protocol-2026-10-03/neural-preparation-data.zip',archive)
with archive.open('rb') as f: assert hashlib.file_digest(f,'sha256').hexdigest()==assets['sha256']
with zipfile.ZipFile(archive) as z: z.extractall(ROOT)
print('Six frozen data files restored; archive checksum verified',flush=True)
if shutil.which('uv') is None: run([sys.executable,'-m','pip','install','uv'])
UV = shutil.which('uv')
run([UV,'python','install','3.11'])
ENV = pathlib.Path('/kaggle/temp/cfpb-neural-env')
if not ENV.exists(): run([UV,'venv','--python','3.11','--seed',str(ENV)])
PY = str(ENV/'bin/python')
run([PY,'-m','pip','install','torch==2.5.1','--index-url','https://download.pytorch.org/whl/cu121'])
run([PY,'-m','pip','install','-r','requirements-neural.txt'],PROJECT)
run([PY,'-c','from src.neural.environment import check_environment; from src.neural.protocol import verify_freeze; print(check_environment()); verify_freeze(); print("ALL FROZEN HASHES VERIFIED")'],PROJECT)
run([PY,'-m','pytest','-q','tests/test_neural_protocol.py','tests/test_neural_statistics.py'],PROJECT)
print('PINNED ENVIRONMENT SETUP COMPLETE',flush=True)
```

```python
# Integration smoke checks, followed immediately by the first real frozen run.
run(['git','fetch','origin'],ROOT)
run(['git','checkout','8af1be6'],ROOT)
LOGS = PROJECT/'outputs/gpu_execution'
LOGS.mkdir(parents=True,exist_ok=True)
for previous in ['smoke_micro_4.log','failure.json','smoke_results.json']:
    p=LOGS/previous
    if p.exists() and not (LOGS/('initial_'+previous)).exists(): shutil.copy2(p,LOGS/('initial_'+previous))
def logged(args, name):
    env=os.environ.copy()
    env.update(PYTHONUNBUFFERED='1', TOKENIZERS_PARALLELISM='false', HF_HOME='/kaggle/temp/cfpb-hf-cache')
    with (LOGS/name).open('w') as f:
        p=subprocess.Popen(args,cwd=PROJECT,env=env,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
        for line in p.stdout:
            print(line,end='',flush=True);f.write(line);f.flush()
        return p.wait()
MICRO=None
for micro in [4,2,1]:
    cmd=[PY,'-m','torch.distributed.run','--standalone','--nproc_per_node=2','-m','src.neural.gpu_smoke','--micro-batch',str(micro)]
    print('STARTING GPU SMOKE',micro,flush=True)
    code=logged(cmd,'smoke_micro_'+str(micro)+'.log')
    if code==0:
        MICRO=micro;break
    text=(LOGS/('smoke_micro_'+str(micro)+'.log')).read_text()
    (LOGS/'failure.json').write_text(json.dumps({'stage':'smoke','micro_batch':micro,'exit_code':code,'future_neural_results_observed':False,'error_tail':text[-12000:]},indent=2))
    if 'out of memory' not in text.lower():
        raise RuntimeError('Genuine smoke blocker; see outputs/gpu_execution/failure.json. No training launched.')
    print('OOM: using the next frozen microbatch/accumulation fallback',flush=True)
assert MICRO is not None, 'Persistent OOM after frozen fallback; stop'
(LOGS/'selected_microbatch.json').write_text(json.dumps({'micro_batch':MICRO,'accumulation':64//(2*MICRO),'global_batch':64,'future_neural_results_observed':False},indent=2))
print('ALL GPU SMOKE CHECKS PASSED; STARTING ACTUAL DEBT COLLECTION LAYA-CE SEED 17',flush=True)
cmd=[PY,'-m','torch.distributed.run','--standalone','--nproc_per_node=2','-m','src.neural.execute_gpu','--domain','debt_collection','--model','laya','--seed','17','--micro-batch',str(MICRO),'--execute-training']
if (PROJECT/'outputs/neural/debt_collection/laya_ce/seed_17/latest.pt').exists(): cmd.append('--resume')
code=logged(cmd,'debt_laya_ce_seed17.log')
assert code==0, 'Training interrupted or failed; preserve latest.pt and inspect the log before resuming'
print('DEBT COLLECTION LAYA-CE SEED 17 COMPLETE',flush=True)

```

```python
# Fixed order only; no inference or performance-based decisions.
import time, datetime
assert json.loads((LOGS/'smoke_results.json').read_text())['passed']
assert (PROJECT/'outputs/neural/debt_collection/laya_ce/seed_17/completion.json').exists(), 'First real run must complete before queue'
models=[('laya','laya_ce'),('matched_head','matched_head'),('modernbert','modernbert'),('laya_rlcd','laya_rlcd')]
started_utc=datetime.datetime.fromisoformat('2026-10-02T19:56:08+00:00').timestamp()
def checkpoint_manifest(folder):
    files=[]
    for name in ['best.pt','latest.pt']:
        p=folder/name
        if p.exists():
            with p.open('rb') as f: digest=hashlib.file_digest(f,'sha256').hexdigest()
            files.append({'file':name,'bytes':p.stat().st_size,'sha256':digest})
    (folder/'checkpoint_manifest.json').write_text(json.dumps({'storage':'Kaggle working output; excluded from normal Git','files':files},indent=2))
checkpoint_manifest(PROJECT/'outputs/neural/debt_collection/laya_ce/seed_17')
for domain in ['debt_collection','credit_card']:
    domain_models=models if domain=='debt_collection' else models[:3]
    for model,key in domain_models:
        for seed in [17,43,101]:
            folder=PROJECT/'outputs/neural'/domain/key/('seed_'+str(seed))
            if (folder/'completion.json').exists():continue
            used=sum(p.stat().st_size for p in pathlib.Path('/kaggle/working').rglob('*') if p.is_file() and not p.is_symlink())
            elapsed=time.time()-started_utc
            if used+9*1024**3>19*1024**3 or elapsed+4*3600>10*3600:
                (LOGS/'queue_resource_pause.json').write_text(json.dumps({'next_domain':domain,'next_model':model,'next_seed':seed,'working_bytes':used,'elapsed_seconds':elapsed,'reason':'Preserve checkpoints within output/session limits; remaining fixed runs await resumed session','future_inference_executed':False},indent=2))
                print('RESOURCE PAUSE BEFORE',domain,model,seed,flush=True)
                raise RuntimeError('Normal resource pause: save Kaggle outputs and resume the fixed queue in a fresh session. No experiment redesign.')
            cmd=[PY,'-m','torch.distributed.run','--standalone','--nproc_per_node=2','-m','src.neural.execute_gpu','--domain',domain,'--model',model,'--seed',str(seed),'--micro-batch',str(MICRO),'--execute-training']
            if (folder/'latest.pt').exists():cmd.append('--resume')
            print('START FIXED RUN',domain,model,seed,flush=True)
            code=logged(cmd,domain+'_'+key+'_seed'+str(seed)+'.log')
            assert code==0, 'Runtime failure: preserve checkpoint and record exact error before resuming'
            checkpoint_manifest(folder)
print('ALL PLANNED TRAINING COMPLETE; NO FUTURE INFERENCE EXECUTED',flush=True)
```
